# Trabalho 01 — Modelos Lineares 2026/2

## Análise da Evolução Temporal dos Registros do Ligue 180 por Meio de Modelos de Regressão Linear

Nesta seção é desenvolvida a análise da evolução mensal dos registros do Ligue 180 no período de janeiro de 2022 a março de 2024.

O objetivo é investigar a existência de uma tendência temporal estatisticamente significativa na quantidade mensal de registros, comparar modelos de regressão linear simples e múltipla e avaliar a adequação do modelo selecionado por meio da análise de resíduos.

## 1. Apresentação, verificação e preparação da base

A base utilizada contém registros do Ligue 180 previamente tratados e unificados em um único DataFrame.

A base consolidada possui **1.299.357 registros e 62 variáveis**, contemplando informações como data de cadastro, canal de atendimento, denunciante, local da ocorrência, frequência da violência, características da vítima e do suspeito, entre outras.

Os registros analisados compreendem o período de **janeiro de 2022 a março de 2024**.

Para o desenvolvimento deste trabalho de Modelos Lineares, serão utilizadas principalmente as seguintes variáveis:

- `DATA_DE_CADASTRO`: data em que o registro foi cadastrado no sistema;
- `MES`: mês e ano do registro, variável derivada de `DATA_DE_CADASTRO`;
- `QUANTIDADE_REGISTROS`: quantidade de registros contabilizados em cada mês;
- `TEMPO`: sequência numérica dos meses, variando de 1 a 27;
- `TRIMESTRE`: trimestre correspondente a cada mês;
- `T_2`, `T_3` e `T_4`: variáveis indicadoras utilizadas para representar os 2º, 3º e 4º trimestres no modelo múltiplo.

O **1º trimestre** será utilizado como categoria de referência no modelo de regressão múltipla.

```python
# Verificar a qualidade e o período da coluna DATA_DE_CADASTRO

print("Tipo da coluna:")
print(df_unificado["DATA_DE_CADASTRO"].dtype)

print("\nQuantidade de datas nulas:")
print(df_unificado["DATA_DE_CADASTRO"].isna().sum())

print("\nMenor data da base:")
print(df_unificado["DATA_DE_CADASTRO"].min())

print("\nMaior data da base:")
print(df_unificado["DATA_DE_CADASTRO"].max())
```

### Verificação da variável temporal

A variável `DATA_DE_CADASTRO` encontra-se no formato adequado para análise temporal.

Não foram identificados valores nulos nessa variável, de modo que todos os registros puderam ser considerados na agregação mensal.

O período disponível na base compreende registros entre janeiro de 2022 e março de 2024.

## 2. Construção da base mensal

Para a aplicação do modelo de regressão linear, os registros serão agrupados por mês de cadastro.

A variável resposta será a **quantidade mensal de registros**, enquanto a variável explicativa será o **tempo**, representado pela sequência dos meses ao longo do período analisado.

```python
# Criar uma base com a quantidade de registros por mês

df_mensal = (
    df_unificado
    .assign(
        MES=lambda x: x["DATA_DE_CADASTRO"].dt.to_period("M")
    )
    .groupby("MES")
    .size()
    .reset_index(name="QUANTIDADE_REGISTROS")
    .sort_values("MES")
    .reset_index(drop=True)
)

# Criar a variável tempo: 1, 2, 3, ..., n
df_mensal["TEMPO"] = range(1, len(df_mensal) + 1)

df_mensal
```

A agregação resultou em 27 observações mensais, correspondentes ao período de janeiro de 2022 a março de 2024.

A variável `QUANTIDADE_REGISTROS` representa o total de registros observados em cada mês, enquanto `TEMPO` representa a sequência cronológica de 1 a 27.

## 3. Hipóteses do trabalho

O objetivo principal é investigar se existe uma tendência linear significativa na quantidade mensal de registros do Ligue 180 ao longo do período analisado.

### Hipótese principal — efeito do tempo

\[
H_0:\beta_1 = 0
\]

**Hipótese nula (H₀):** não existe relação linear estatisticamente significativa entre o tempo e a quantidade mensal de registros.

\[
H_1:\beta_1 \neq 0
\]

**Hipótese alternativa (H₁):** existe relação linear estatisticamente significativa entre o tempo e a quantidade mensal de registros.

Será adotado nível de significância de 5% (\(\alpha = 0,05\)).

Além da tendência temporal, será investigado se a inclusão dos trimestres acrescenta capacidade explicativa ao modelo.

Para essa análise:

\[
H_0:\beta_{T2}=\beta_{T3}=\beta_{T4}=0
\]

A inclusão dos trimestres não acrescenta efeito significativo ao modelo.

\[
H_1:\text{pelo menos um dos coeficientes dos trimestres é diferente de zero}
\]

Pelo menos um trimestre apresenta efeito adicional sobre a quantidade mensal de registros, após considerar a tendência temporal.

## 4. Análise exploratória

O gráfico de dispersão permitirá observar visualmente se existe indício de uma relação linear entre o tempo e a quantidade de registros.

Entretanto, a correlação mede apenas a intensidade e a direção da associação linear entre as variáveis e não estabelece relação de causa e efeito.

## 5. Preparação das variáveis para a regressão múltipla

Para incluir o trimestre no modelo de regressão, serão utilizadas variáveis indicadoras (dummies).

O **1º trimestre** será utilizado como categoria de referência. Dessa forma, serão criadas três variáveis:

- `T_2`: indica observações do 2º trimestre;
- `T_3`: indica observações do 3º trimestre;
- `T_4`: indica observações do 4º trimestre.

Quando todas essas variáveis forem iguais a zero, a observação pertence ao 1º trimestre.

```python
# Criar a variável trimestre a partir da coluna MES

df_mensal["TRIMESTRE"] = df_mensal["MES"].dt.quarter

df_mensal[["MES", "QUANTIDADE_REGISTROS", "TEMPO", "TRIMESTRE"]]
```

```python
import pandas as pd

# Criar variáveis dummies para os trimestres
dummies_trimestre = pd.get_dummies(
    df_mensal["TRIMESTRE"],
    prefix="T",
    drop_first=True,
    dtype=int
)

# Juntar as dummies à base mensal
df_modelo = pd.concat(
    [df_mensal, dummies_trimestre],
    axis=1
)

df_modelo[
    ["MES", "QUANTIDADE_REGISTROS", "TEMPO", "TRIMESTRE", "T_2", "T_3", "T_4"]
]
```

## 6. Modelo de regressão linear simples

Inicialmente, será ajustado um modelo de regressão linear simples para avaliar a relação entre o tempo e a quantidade mensal de registros.

A variável resposta será `QUANTIDADE_REGISTROS` e a variável explicativa será `TEMPO`.

O modelo pode ser representado por:

\[
Y_i = \beta_0 + \beta_1 X_i + \varepsilon_i
\]

onde:

- \(Y_i\) representa a quantidade mensal de registros;
- \(X_i\) representa o tempo em meses;
- \(\beta_0\) é o intercepto;
- \(\beta_1\) representa a variação média esperada na quantidade de registros para cada mês adicional;
- \(\varepsilon_i\) representa o erro aleatório.

```python
import statsmodels.api as sm

# Definir a variável explicativa (X) e a variável resposta (Y)

X_simples = df_modelo[["TEMPO"]]
Y = df_modelo["QUANTIDADE_REGISTROS"]

# Adicionar a constante do modelo (intercepto)
X_simples = sm.add_constant(X_simples)

# Ajustar o modelo de regressão linear simples
modelo_simples = sm.OLS(Y, X_simples).fit()

# Exibir o resumo do modelo
print(modelo_simples.summary())
```

### Interpretação do modelo de regressão linear simples

O modelo estimado foi:

\[
\hat{Y} = 35.251,03 + 919,52 \times TEMPO
\]

O coeficiente associado à variável `TEMPO` foi de aproximadamente **919,52**, indicando que, no período analisado, o avanço de um mês está associado a um aumento médio estimado de aproximadamente **920 registros mensais**.

O coeficiente apresentou p-valor inferior a 0,001 e intervalo de confiança de 95% entre aproximadamente **706 e 1.133 registros por mês**. Assim, rejeita-se a hipótese nula \(H_0:\beta_1=0\), encontrando-se evidências estatísticas de uma tendência temporal linear positiva.

O modelo apresentou \(R^2 = 0,759\), indicando que aproximadamente **75,9% da variabilidade observada na quantidade mensal de registros é explicada pela tendência temporal linear**. O \(R^2\) ajustado foi de **0,749**.

A estatística F foi de aproximadamente **78,72**, com p-valor de **3,39 × 10⁻⁹**, indicando significância estatística global do modelo.

O intercepto estimado corresponde ao valor previsto quando `TEMPO = 0`, condição situada fora do período efetivamente observado. Por isso, sua interpretação prática é limitada.

Os resultados representam associação estatística e não permitem estabelecer relação causal.

## 7. Modelo de regressão linear múltipla

Após a análise do modelo simples, será ajustado um modelo de regressão múltipla incluindo, além do tempo, variáveis indicadoras para os trimestres do ano.

O objetivo é verificar se a inclusão das variáveis referentes aos trimestres acrescenta capacidade explicativa à tendência temporal já identificada.

O 1º trimestre será utilizado como categoria de referência.

O modelo pode ser representado por:

\[
Y_i = \beta_0 + \beta_1 TEMPO_i + \beta_2 T2_i + \beta_3 T3_i + \beta_4 T4_i + \varepsilon_i
\]

onde:

- \(Y_i\) representa a quantidade mensal de registros;
- `TEMPO` representa a sequência dos meses;
- `T_2`, `T_3` e `T_4` representam os trimestres 2, 3 e 4;
- o 1º trimestre é a categoria de referência;
- \(\varepsilon_i\) representa o erro aleatório.

```python
# Definir as variáveis explicativas do modelo múltiplo

X_multiplo = df_modelo[
    ["TEMPO", "T_2", "T_3", "T_4"]
]

Y = df_modelo["QUANTIDADE_REGISTROS"]

# Adicionar a constante do modelo
X_multiplo = sm.add_constant(X_multiplo)

# Ajustar o modelo de regressão múltipla
modelo_multiplo = sm.OLS(Y, X_multiplo).fit()

# Exibir o resumo do modelo
print(modelo_multiplo.summary())
```

### Interpretação do modelo de regressão múltipla

No modelo múltiplo, o coeficiente da variável `TEMPO` foi de aproximadamente **918,63**, mantendo-se muito próximo do valor encontrado no modelo de regressão simples.

Isso indica que, mesmo após considerar possíveis diferenças entre os trimestres, cada mês adicional está associado a um aumento médio estimado de aproximadamente **919 registros mensais**.

O p-valor associado à variável `TEMPO` foi inferior a 0,05, indicando que a tendência temporal permanece estatisticamente significativa.

O 1º trimestre foi utilizado como categoria de referência. A variável T_2 apresentou coeficiente de -866,44 e p-valor de 0,698; T_3, coeficiente de -4.070,67 e p-valor de 0,075; e T_4, coeficiente de -785,39 e p-valor de 0,725. Como os três p-valores são superiores a 0,05, não foram encontradas evidências estatísticas de diferenças entre o primeiro trimestre e os demais, após considerar o efeito do tempo.

O modelo apresentou \(R^2 = 0,794\), indicando que aproximadamente **79,4% da variação na quantidade mensal de registros é explicada conjuntamente pelo tempo e pelos trimestres**.

O \(R^2\) ajustado foi de **0,757**, pouco superior ao observado no modelo simples (0,749). Isso indica que a inclusão das variáveis de trimestre trouxe apenas uma pequena melhora na capacidade explicativa do modelo.

De forma geral, os resultados sugerem que a **a tendência temporal permanece como o principal efeito identificado entre as variáveis analisadas**, enquanto não foram identificadas diferenças estatisticamente significativas entre os trimestres ao nível de 5%.

A avaliação formal da contribuição conjunta das variáveis de trimestre será realizada por meio da comparação entre os modelos simples e múltiplo.

### Comparação formal entre os modelos

Após o ajuste dos modelos simples e múltiplo, será realizada uma comparação formal para verificar se a inclusão das variáveis indicadoras dos trimestres proporciona melhora estatisticamente significativa no ajuste.

Como o modelo simples está contido no modelo múltiplo, será utilizado o teste F para modelos aninhados.

```python
from statsmodels.stats.anova import anova_lm

comparacao_modelos = anova_lm(
    modelo_simples,
    modelo_multiplo
)

comparacao_modelos
```

### Interpretação da comparação

O teste F para comparação entre os modelos apresentou estatística de aproximadamente **1,248** e p-valor de **0,3165**.

Como o p-valor é superior ao nível de significância de 5%, **não se rejeita a hipótese nula de que os coeficientes adicionais associados aos trimestres sejam conjuntamente iguais a zero**. Assim, não foram encontradas evidências estatísticas de que a inclusão das variáveis referentes aos trimestres produza melhora significativa no ajuste em relação ao modelo simples.

Além disso, embora o \(R^2\) tenha aumentado de **0,759 para 0,794**, o \(R^2\) ajustado apresentou apenas pequena variação, passando de **0,749 para 0,757**.

Dessa forma, o modelo de regressão linear simples foi considerado **mais parcimonioso** para representar a relação analisada, uma vez que utiliza menor número de parâmetros sem perda estatisticamente significativa de capacidade explicativa.

## 8. Diagnóstico do modelo selecionado — Regressão Linear Simples

Após a comparação entre os modelos, o modelo de regressão linear simples foi selecionado como a representação mais parcimoniosa da relação analisada.

Nesta etapa serão avaliados os resíduos desse modelo, considerando linearidade, normalidade, homocedasticidade, independência e presença de observações influentes.

```python
# Obter valores ajustados e resíduos do modelo simples

valores_previstos_simples = modelo_simples.fittedvalues
residuos_simples = modelo_simples.resid

plt.figure(figsize=(10, 6))

plt.scatter(
    valores_previstos_simples,
    residuos_simples
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Valores ajustados")
plt.ylabel("Resíduos")
plt.title("Resíduos versus valores ajustados — modelo linear simples")

plt.grid(True, alpha=0.3)

# Salvar em alta resolução
plt.savefig(
    "03_residuos_valores_ajustados.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Interpretação dos resíduos versus valores ajustados

Os resíduos encontram-se distribuídos tanto acima quanto abaixo da linha zero. Entretanto, observa-se certa estrutura em sua distribuição, com maior presença de resíduos positivos nas extremidades dos valores ajustados e concentração de resíduos negativos em parte da região intermediária.

Esse comportamento sugere a possibilidade de uma curvatura não capturada integralmente pelo modelo linear simples, indicando que a relação entre o tempo e a quantidade mensal de registros pode não ser perfeitamente linear ao longo de todo o período.

Por outro lado, não se observa visualmente um padrão claro de aumento ou redução progressiva da dispersão dos resíduos, como o formato de funil normalmente associado à heterocedasticidade.

Dessa forma, o modelo simples capta a tendência geral de crescimento, mas a suposição de linearidade deve ser interpretada com cautela.

### Verificação da normalidade dos resíduos

A normalidade dos resíduos será avaliada inicialmente por meio de dois recursos gráficos:

- histograma dos resíduos;
- gráfico Q-Q Plot.

Em um modelo adequado, espera-se que os resíduos apresentem distribuição aproximadamente simétrica em torno de zero e que, no Q-Q Plot, os pontos permaneçam próximos da linha de referência.

```python
plt.figure(figsize=(10, 6))

plt.hist(
    residuos_simples,
    bins=8,
    edgecolor="black"
)

plt.xlabel("Resíduos")
plt.ylabel("Frequência")
plt.title("Distribuição dos resíduos — modelo linear simples")

plt.grid(True, alpha=0.3)

plt.savefig(
    "04_histograma_residuos.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
sm.qqplot(
    residuos_simples,
    line="45",
    fit=True
)

plt.title("Q-Q Plot dos resíduos — modelo linear simples")

plt.savefig(
    "05_qqplot_residuos.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Interpretação gráfica da normalidade

O histograma apresenta alguma irregularidade na distribuição dos resíduos, o que deve ser interpretado considerando o número reduzido de observações mensais (\(n=27\)).

No Q-Q Plot, a maior parte dos pontos permanece relativamente próxima da linha de referência, embora sejam observados pequenos desvios, principalmente em algumas regiões da distribuição.

A avaliação visual, portanto, não evidencia uma violação acentuada da normalidade. Para complementar essa análise, foi aplicado o teste de Shapiro-Wilk.

```python
from scipy.stats import shapiro

estatistica_shapiro, p_valor_shapiro = shapiro(residuos_simples)

print(f"Estatística de Shapiro-Wilk: {estatistica_shapiro:.4f}")
print(f"P-valor: {p_valor_shapiro:.4f}")
```

### Resultado do teste de normalidade

O teste de Shapiro-Wilk apresentou estatística de 0,9528 e p-valor de 0,2509.

Considerando o nível de significância de 5%, o p-valor obtido é superior a 0,05. Dessa forma, não há evidência estatística suficiente para rejeitar a hipótese de normalidade dos resíduos.

Esse resultado, analisado em conjunto com o Q-Q Plot e o histograma, indica que a suposição de normalidade é razoavelmente compatível com os resíduos do modelo.

### Verificação da homocedasticidade

Além da análise gráfica dos resíduos, será utilizado o teste de Breusch-Pagan para verificar se a variância dos erros permanece aproximadamente constante.

As hipóteses do teste são:

- **H₀:** os resíduos apresentam variância constante (homocedasticidade);
- **H₁:** os resíduos apresentam variância não constante (heterocedasticidade).

Será adotado nível de significância de 5%.

```python
from statsmodels.stats.diagnostic import het_breuschpagan

bp_teste = het_breuschpagan(
    residuos_simples,
    modelo_simples.model.exog
)

estatistica_lm = bp_teste[0]
p_valor_lm = bp_teste[1]

print(f"Estatística de Breusch-Pagan: {estatistica_lm:.4f}")
print(f"P-valor: {p_valor_lm:.4f}")
```

### Resultado do teste de homocedasticidade

O teste de Breusch-Pagan apresentou estatística de **0,3428** e p-valor de **0,5582**.

Como o p-valor é superior ao nível de significância de 5% (\(\alpha = 0,05\)), não rejeitamos a hipótese nula de variância constante dos resíduos.

Assim, não foram encontradas evidências estatísticas de heterocedasticidade.

Esse resultado está de acordo com a análise visual do gráfico de resíduos versus valores ajustados, no qual também não foi observado um padrão evidente de aumento ou redução da dispersão dos resíduos.

```python
plt.figure(figsize=(10, 6))

plt.plot(
    df_modelo["TEMPO"],
    residuos_simples,
    marker="o"
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.xlabel("Tempo (meses)")
plt.ylabel("Resíduos")
plt.title("Resíduos ao longo do tempo — modelo linear simples")

plt.grid(True, alpha=0.3)

plt.savefig(
    "06_residuos_tempo.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
from statsmodels.stats.stattools import durbin_watson

dw = durbin_watson(residuos_simples)

print(f"Estatística de Durbin-Watson: {dw:.4f}")
```

### Verificação da independência dos resíduos

O gráfico dos resíduos ao longo do tempo revela oscilações em torno de zero, mas também apresenta alguns períodos consecutivos em que os resíduos permanecem predominantemente com o mesmo sinal. Esse comportamento merece atenção por se tratar de observações ordenadas temporalmente.

A estatística de Durbin-Watson foi de 1,5097. Valores próximos de 2 são geralmente associados à ausência de autocorrelação de primeira ordem, enquanto valores inferiores a 2 podem sugerir autocorrelação positiva.

Assim, o resultado obtido sugere a possibilidade de alguma dependência temporal entre os resíduos, embora não seja suficiente, isoladamente, para caracterizar uma violação definitiva da independência.

Essa ressalva é particularmente relevante neste estudo porque as observações correspondem a meses consecutivos. Portanto, a independência dos erros deve ser considerada um ponto de atenção na interpretação inferencial do modelo.

### Verificação de pontos influentes

Por fim, será avaliada a presença de observações potencialmente influentes sobre o ajuste do modelo. Para isso, será utilizada a Distância de Cook, que permite avaliar o impacto de cada observação sobre as estimativas da regressão. Como referência, será utilizada a regra prática \(4/n\).” O código abaixo está correto e deve permanecer.

```python
# Distância de Cook do modelo simples

influencia_simples = modelo_simples.get_influence()

cook_dist_simples = influencia_simples.cooks_distance[0]

df_modelo["COOK_DISTANCE_SIMPLES"] = cook_dist_simples

limite_cook = 4 / len(df_modelo)

print(f"Limite de referência 4/n: {limite_cook:.4f}")

df_modelo[
    ["MES", "QUANTIDADE_REGISTROS", "COOK_DISTANCE_SIMPLES"]
].sort_values(
    "COOK_DISTANCE_SIMPLES",
    ascending=False
).head(10)
```

### Interpretação da Distância de Cook

Como referência, foi utilizada a regra prática de \(4/n\). Considerando as 27 observações do modelo:

\[
\frac{4}{27} \approx 0,148
\]

Março de 2024 apresentou Distância de Cook de aproximadamente **0,183**, valor superior ao limite de referência de **0,148**, sendo identificado como uma observação potencialmente influente.

Esse resultado não significa que a observação deva ser automaticamente excluída. Março de 2024 corresponde a uma observação válida da série e não foi identificada evidência de erro nos dados.

Por esse motivo, a observação será mantida no modelo principal e sua influência será avaliada por meio de uma análise de sensibilidade.

### Análise de sensibilidade — março de 2024

Para verificar se março de 2024 altera a conclusão do modelo, a regressão linear simples será reestimada excluindo temporariamente essa observação.

A exclusão será realizada apenas para fins de comparação. A observação continuará sendo mantida no modelo principal.

```python
# Análise de sensibilidade: modelo sem março de 2024

df_sem_mar2024 = df_modelo[
    df_modelo["MES"].astype(str) != "2024-03"
].copy()

X_sem_mar2024 = sm.add_constant(
    df_sem_mar2024[["TEMPO"]]
)

Y_sem_mar2024 = df_sem_mar2024["QUANTIDADE_REGISTROS"]

modelo_sem_mar2024 = sm.OLS(
    Y_sem_mar2024,
    X_sem_mar2024
).fit()

print("Modelo original")
print(f"Coeficiente de TEMPO: {modelo_simples.params['TEMPO']:.2f}")
print(f"R²: {modelo_simples.rsquared:.4f}")
print(f"P-valor de TEMPO: {modelo_simples.pvalues['TEMPO']:.10e}")

print("\nModelo sem março/2024")
print(f"Coeficiente de TEMPO: {modelo_sem_mar2024.params['TEMPO']:.2f}")
print(f"R²: {modelo_sem_mar2024.rsquared:.4f}")
print(f"P-valor de TEMPO: {modelo_sem_mar2024.pvalues['TEMPO']:.10e}")
```

### Interpretação da análise de sensibilidade

A retirada de março de 2024 reduziu o coeficiente associado ao tempo de aproximadamente **919,52 para 865,74 registros por mês**, correspondendo a uma redução de cerca de 5,8% na magnitude estimada do efeito temporal.

O \(R^2\) passou de **0,7590 para 0,7325**.

Apesar dessas alterações, o coeficiente de `TEMPO` permaneceu positivo e estatisticamente significativo, com p-valor de aproximadamente **2,50 × 10⁻⁸**.

Portanto, março de 2024 exerce influência sobre a magnitude das estimativas, mas sua retirada não altera a conclusão central quanto à existência de uma tendência temporal positiva e estatisticamente significativa.

Como a observação é válida, ela permanece no modelo principal.

## 10. Conclusão

O trabalho teve como objetivo analisar a evolução mensal dos registros do Ligue 180 entre janeiro de 2022 e março de 2024, avaliando a existência de uma tendência temporal por meio de modelos de regressão linear.

A análise exploratória indicou uma associação linear positiva forte entre o tempo e a quantidade mensal de registros, com correlação de **0,8712**.

No modelo de regressão linear simples, estimou-se um aumento médio de aproximadamente **920 registros por mês**, com \(R^2 = 0,759\). O coeficiente temporal apresentou significância estatística, permitindo rejeitar a hipótese nula de ausência de relação linear entre o tempo e a quantidade mensal de registros.

Posteriormente, foi ajustado um modelo de regressão linear múltipla incluindo variáveis indicadoras referentes aos trimestres do ano. Embora o \(R^2\) tenha aumentado para **0,794**, o teste F de comparação entre os modelos apresentou p-valor de **0,3165**, não fornecendo evidências de melhora estatisticamente significativa com a inclusão conjunta dos trimestres.

Dessa forma, em relação à hipótese complementar, não se rejeita a hipótese nula de que os efeitos adicionais dos trimestres sejam conjuntamente iguais a zero.

O modelo de regressão linear simples foi selecionado como mais parcimonioso e mostrou-se adequado para responder às hipóteses estabelecidas neste estudo.

A análise dos resíduos não apresentou evidências de violação das suposições de normalidade e homocedasticidade. Entretanto, foram identificados pontos de atenção relacionados à possível não linearidade, à dependência temporal dos resíduos e à influência de março de 2024.

A análise de sensibilidade demonstrou que a retirada de março de 2024 reduz a magnitude do coeficiente temporal, mas não modifica sua direção nem sua significância estatística. Assim, a conclusão sobre a tendência temporal positiva não depende exclusivamente dessa observação.

Portanto, o modelo simples oferece evidências estatísticas consistentes de crescimento da quantidade mensal de registros ao longo do período analisado, embora as limitações identificadas devam ser consideradas na interpretação dos resultados.

Por fim, ressalta-se que o objeto analisado corresponde à **quantidade de registros realizados no Ligue 180**, e não diretamente à ocorrência real de violência contra as mulheres. Assim, o crescimento dos registros não permite concluir, isoladamente, que houve aumento da violência.
